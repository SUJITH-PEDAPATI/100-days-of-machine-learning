```python
import pandas as pd
import numpy as np

data_url = "http://lib.stat.cmu.edu/datasets/boston"
raw_df = pd.read_csv(data_url, sep="\s+", skiprows=22, header=None)
data = np.hstack([raw_df.values[::2, :], raw_df.values[1::2, :2]])
target = raw_df.values[1::2, 2]
```

```python
data
```

```python
df = pd.DataFrame(data)
```

```python
df
```

```python
x = df
y = target

print(f"Shape of x: {x.shape}")
print(f"Shape of y: {y.shape}")
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.svm import SVR
from sklearn.model_selection import cross_val_score
```

```python
lr = LinearRegression()
dt = DecisionTreeRegressor()
svr = SVR()
```

```python
estimators = [('lr',lr),('dt',dt),('svr',svr)]
```

```python
for estimator in estimators:
  scorees = cross_val_score(estimator[1],x,y,cv = 10,scoring = 'r2')
  print(f"{estimator[0]} : {np.mean(scorees)}")
```

```python
from sklearn.ensemble import VotingRegressor
```

```python
vr = VotingRegressor(estimators = estimators)
```

```python
scoring = cross_val_score(vr,x,y,cv = 10,scoring = 'r2')
```

```python
print("Voting Regressor", np.round(np.mean(scoring),2))
```

```python
for i in range(1,4):
  for j in range(1,4):
    for k in range(1,4):
      vr = VotingRegressor(estimators,weights = [i,j,k])
      scores = cross_val_score(vr,x,y,cv = 10,scoring = 'r2')
      print("For i = {}, j = {}, k = {}".format(i,j,k),np.round(np.mean(scores),2))
```

```python
dt1 = DecisionTreeRegressor(max_depth = 1)
dt2 = DecisionTreeRegressor(max_depth = 3)
dt3 = DecisionTreeRegressor(max_depth = 5)
dt4 = DecisionTreeRegressor(max_depth = 7)
dt5 = DecisionTreeRegressor(max_depth = 9)
dt6 = DecisionTreeRegressor(max_depth = None)
```

```python
estimators = [('dt1',dt1),('dt2',dt2),('dt3',dt3),('dt4',dt4),('dt5',dt5),('dt6',dt6)]
```

```python
for estimator in estimators:
  scorees = cross_val_score(estimator[1],x,y,cv = 10,scoring = 'r2')
  print(f"{estimator[0]} : {np.mean(scorees)}")
```

```python
vr = VotingRegressor(estimators = estimators)
scores = cross_val_score(vr,x,y,cv = 10,scoring = 'r2')
print("Voting Regressor", np.round(np.mean(scores),2))
```
